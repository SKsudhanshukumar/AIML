```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris=pd.read_csv('Datasets/iris.csv')
iris.head()
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
# cmap and alpha
```

Plot size

```python
plt.figure(figsize=(10,7))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

Annotations

```python
batter=pd.read_csv('Datasets/batter.csv')
```

```python
sample_df=batter.head(100).sample(25,random_state=5)
```

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

Horizontal and Vertical lines

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
plt.axhline(130,color='red')
plt.axvline(30,color='red')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

Subplots

```python
# A diff way to plot graphs
fig, ax=plt.subplots(figsize=(10,8))
ax.scatter(batter['avg'],batter['strike_rate'])
ax.set_title('Performance')
ax.set_xlabel('Avarage')
ax.set_ylabel('Strike Rate')
fig.show()
```

```python
# plotting multiple graph simultaneously
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True)
ax[0].scatter(batter['avg'],batter['strike_rate'])
ax[1].scatter(batter['avg'],batter['runs'])
ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Avg Vs Strike Rate')

ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2)
ax[0,0].scatter(batter['avg'],batter['strike_rate'])
ax[0,1].scatter(batter['avg'],batter['runs'])
ax[1,0].hist(batter['avg'])
ax[1,1].hist(batter['runs'])
```

```python
fig = plt.figure()

ax1=fig.add_subplot(2,1,1)
ax1.scatter(batter['avg'],batter['strike_rate'])

ax2=fig.add_subplot(2,1,2)
ax2.hist(batter['runs'])
```

3D Scatter Plots

```python
fig = plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])
ax.set_title('IPL batsman analysis')
ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
```

3D Surface Plot

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y)
```

```python
z=xx**2 + yy**2
```

```python
fig =plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z=np.sin(xx)+np.cos(yy)
fig =plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
z=xx**2 + yy**2
fig =plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig =plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig =plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

Hitmap

```python
delivery=pd.read_csv('Datasets/IPL_Ball_by_Ball_2008_2022.csv')
```

```python
temp_df=delivery[((delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6))]
```

```python
grid=temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
```

```python
plt.figure(figsize=(5,5))
plt.imshow(grid)
plt.colorbar()
plt.yticks(delivery['overs'].unique(),list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.show()
```
