2D Line plot

• Bivariate Analysis
• Categorical -> numerical
• Use case - Time series data

```python
# plotting a simple function
price =[48000,54000,57000,49000,47000,45000]
year =[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
batsman=pd.read_csv('Datasets/sharma-kohli.csv')
batsman
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma Vs Virat Kohli career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# color(hex)
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')
plt.title('Rohit Sharma Vs Virat Kohli career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# line(width and style)
plt.plot(batsman['index'],batsman['V Kohli'],linestyle='dotted',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dashdot',linewidth=2)
plt.title('Rohit Sharma Vs Virat Kohli career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# markers
plt.plot(batsman['index'],batsman['V Kohli'],marker='D',markersize=7)
plt.plot(batsman['index'],batsman['RG Sharma'],marker='D')
plt.title('Rohit Sharma Vs Virat Kohli career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend 
plt.plot(batsman['index'],batsman['V Kohli'],label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],label='Rohit')
plt.title('Rohit Sharma Vs Virat Kohli career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend(loc='best')
```

```python
# limitting axes
price =[48000,54000,57000,49000,47000,4500000]
year =[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
plt.ylim(40000,75000)
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma Vs Virat Kohli career comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.grid()
```

Bar Chart

• Bivariate Analysis
• numerical vs Categorical
• Use case- Aggregate analysis of groups

```python
# simple bar chart
children= [10,20,40,10,30]
colors=['red','blue','green','yellow','pink']
plt.bar(colors,children,color='black')
```

```python
# bar chart using data
df=pd.read_csv('Datasets/batsman_season_record.csv')
plt.bar(df['batsman'],df['2015'])
```

```python
# horizontal bar chat
plt.barh(colors,children,color='black')
```

```python
# color and label
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='blue',label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='orange',label='2016')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='green',label='2017')
plt.legend()
plt.show()
```

```python
# Multiple bar charts
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='blue',label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='orange',label='2016')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='green',label='2017')
plt.legend()
plt.show()
```

```python
# xtrick
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='blue',label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='orange',label='2016')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='green',label='2017')
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.legend()
plt.show()
```

```python
# problem
children= [10,20,40,10,30]
colors=[' red red red red red','blue blue blue blue blue','green green green green','yellow yellow yellow yellow yellow','pink pink pink pink']
plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
```

```python
# Stacked bar chart
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=(df['2016']+df['2015']),label='2017')
plt.legend()
```

Pie Chart

• Univariate/Bivariate Analysis
• Categorical vs numerical
• Use case - To find contribution on a standard scale

```python
# simple data
data=[23,45,100,20,40]
subjects=['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)
plt.show()
```

```python
# datasets
df=pd.read_csv('Datasets/gayle-175.csv')
plt.pie(df['batsman_runs'],labels=df['batsman'])
plt.show()
```

```python
# percentage
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
```

```python
# explode
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0])
plt.show()
```
