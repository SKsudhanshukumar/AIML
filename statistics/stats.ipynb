### Measure of dispersion
1. Variance
2. Standard Deviation

```python
statistics.variance(ages)
```

```python
np.var(ages,axis=0)
```

```python
# sample variance (n)
def variance(data):
    n=len(ages)
    mean=sum(data)/n

    deviation=[(x-mean)**2 for x in data]
    variance=sum(deviation)/(n)
    return variance
```

```python
# population variance (n-1)
def variance(data,dof=0):
    n=len(ages)
    mean=sum(data)/n

    deviation=[(x-mean)**2 for x in data]
    variance=sum(deviation)/(n-dof)
    return variance
```

```python
variance(ages,dof=1)
```

```python
statistics.pvariance(ages)
```

```python
import math
```

```python
# standard deviation
math.sqrt(statistics.pvariance(ages))
```

### Check Whether Disstribution is Normal Distribution

```python
# If you Want to Check whether feature is guassian or normal distributed
# Q-Q plot
import matplotlib.pyplot as plt
import scipy.stats as stat
import pylab
def plot_data(sample):
    plt.figure(figsize=(10,6))
    plt.subplot(1,2,1)
    sns.histplot(sample)
    plt.subplot(1,2,2)
    stat.probplot(sample,dist='norm',plot=pylab)
    plt.show()
```

```python
# Create a normal distributed Data
s=np.random.normal(0.5,0.2,1000)
plot_data(s)
```

```python
plot_data(df['sepal_width'])
```
