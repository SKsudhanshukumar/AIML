ARRAY OPERATIONS

```python
# SCALAR OPERATIONS
# Arithmatice 
print(a1+4)
print(a1*2)
```

ARRAY FUNCTIONS

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
# max/min/sum/product
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
```

```python
# mean/median/std/var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
```

```python
# Trignometric Functions
np.sin(a1)
```

```python
# Dot Product
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
print(a2)
print(a3)
print(np.dot(a2,a3))
```

Stacking

```python
a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)
```

```python
# Horizontal stacking
np.hstack((a4,a5))
```

```python
# Vertical stacking
np.vstack((a4,a5))
```

Splitting

```python
# Horizontal splitting
print(a4)
np.hsplit(a4,2)
```

```python
# Vertical splitting
print(a4)
np.vsplit(a5,3)
```
