```python
import numpy as np
import pandas as pd
```

SERIES FROM LISTS

```python
# string
country = ['India', 'England', 'USA', 'Nepal', 'Sri Lanka']
pd.Series(country)
```

```python
# Integers
runs=[13,34,23,67,89]
runs_ser=pd.Series(runs)
```

```python
# Custom index
marks=[67,57,89,100]
subject=['maths','english','science','hindi']
pd.Series(marks,index=subject)
```

```python
# Setting a name
pd.Series(marks,index=subject,name="sudhanshu ke marks")
```

SERIES FROM DICTIONARY

```python
marks={
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}
marks_series=pd.Series(marks,name='sudhanshu ke number')
marks_series
```
